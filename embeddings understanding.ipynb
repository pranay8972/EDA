***1.WHAT IS EMBEDDINGS?***


- Embedings are numerical repreentations of text,images,audio, or any data. For text(usually our focus), and embedding is:

    - A vector(long list of numbers)
    - Generated by a model
    - Designed so that similar meanings-> similar numbers

***2.WHY USE EMBEDDINGS?***

- Because computers don't understand text directly- they iunderstand numbers

- Embeddings allows system to:

    - search semantically("meaning based search")
    - Build recommendation systems
    - measure similarity between texts
    - Cluster documents
    - RAG(Retrieval-Augmented Generation),including chatgroq workflows


***3.WHAT IS A SIMILARITYY SCORE?***

- once you have embeddings, you measure similarity using a mathematical formula
- The mosr common one is:
**cosine sililarity**
it measures the angle between two vectors
    - 1.0-> identical meaning
    - 0.0-> unrwelated
    - 1.0-> opposite meaning

```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
import warnings
warnings.filterwarnings('ignore')
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentences = [
    "I like apples.",
    "I enjoy eating apples.",
    "I drive a motorcycle."
]
    
```

```python
embeddings = model.encode(sentences)
```
