```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["John", "Priya", "Arjun", "Neha"],
    "city": ["Mumbai", "Delhi", "Chennai", "Kolkata"]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 3, 1, 5],
    "amount": [500, 650, 300, 800]
})

customers,orders
```

```python
inner_join=pd.merge(customers,orders,on='customer_id',how='inner')
inner_join
```

```python
left_join=pd.merge(customers,orders,on='customer_id',how='left')
left_join
```

```python
right_join=pd.merge(customers,orders,on='customer_id',how='right')
right_join
```

```python
customers_no_order = customers[~customers['customer_id'].isin(orders['customer_id'])]
customers_no_order
```

```python
not_in_cust_table= orders[~orders['customer_id'].isin(customers['customer_id'])]
not_in_cust_table
```
